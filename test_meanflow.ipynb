# MeanFlow Adapter Testing

This notebook demonstrates how to use the MeanFlow adapter for model creation, training, and sampling.

## 1. Import Required Libraries

```python
import torch
from unified_model import ModelFactory
```

## 2. Create MeanFlow Model

Configure model parameters and create a MeanFlow model instance using ModelFactory.

```python
# Model configuration
config = {
    'img_resolution': 32,
    'in_channels': 3,
    'out_channels': 3,
    'dropout': 0.2,
    'ratio': 0.75,
    'ema_decay': 0.9999,
}

# Create model
model = ModelFactory.create_model('meanflow', config)
print("✓ MeanFlow model created successfully")
```

## 3. Test Forward Pass

Test the model's forward propagation and loss calculation using random input.

```python
# Create random input
batch_size = 2
x = torch.randn(batch_size, 3, 32, 32)
aug_cond = None

# Forward pass
z, logdet, other = model.forward(x, aug_cond)

print(f"✓ Forward pass successful")
print(f"  Input shape: {x.shape}")
if 'loss' in other:
    print(f"  Loss value: {other['loss']:.4f}")
```

## 4. Test Sampling

Generate samples from the model and check the output.

```python
# Generate samples
num_samples = 2
samples = model.sample(
    num_samples=num_samples, 
    device='cpu',
    input_shape=(3, 32, 32)
)

print(f"✓ Sampling successful")
print(f"  Generated samples shape: {samples.shape}")
print(f"  Sample value range: [{samples.min():.4f}, {samples.max():.4f}]")
```

## 5. Visualize Generated Samples (Optional)

Use matplotlib to visualize the generated samples.

```python
import matplotlib.pyplot as plt
import numpy as np

# Normalize samples to [0, 1] range
def normalize_image(img):
    img = img - img.min()
    img = img / (img.max() + 1e-8)
    return img

# Visualize
fig, axes = plt.subplots(1, num_samples, figsize=(10, 5))
if num_samples == 1:
    axes = [axes]

for i in range(num_samples):
    img = samples[i].detach().cpu().numpy().transpose(1, 2, 0)
    img = normalize_image(img)
    axes[i].imshow(img)
    axes[i].axis('off')
    axes[i].set_title(f'Sample {i+1}')

plt.suptitle('MeanFlow Generated Samples')
plt.tight_layout()
plt.show()
```
